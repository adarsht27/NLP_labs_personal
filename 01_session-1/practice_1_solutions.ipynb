## 3. REBUILD
Fill each `...`. The `assert` goes silent when you are right.

```
n_docs = len(df)
assert n_docs == len(df)
```

```
words_per_doc = df['doc'].str.split().str.len()
assert words_per_doc.iloc[2] == len(df['doc'][2].split())
```

```
longest = df.loc[df['doc'].str.len().idxmax(), 'doc']
print(longest)
assert longest in set(df['doc'])
```

```
# Compare with your PREDICT guesses
print(n_docs, words_per_doc.mean().round(1), len(longest))
```
